### 👉 Finger-Nibbling Exercise 2

Let's see what happens when we make our computer faster.

1. Try **32, 64, 128, and 256** bits.
2. Run the cell after each change and compare the time required.
3. Now make `guesses_per_second` **1,000× faster** and run it again.

**Think about it:** Does making the computer 1,000× faster solve the problem of a large keyspace?

There’s nothing to submit. Just change things, run the code, and see what happens.

```python
bits = 64
guesses_per_second = 1_000_000_000_000

seconds = (2 ** bits) / guesses_per_second
years = seconds / (60 * 60 * 24 * 365)

print(f"{bits}-bit secret")
print(f"Trying {guesses_per_second:,} guesses/second")
print(f"Time to try every possibility: {years:,.2f} years")
```

### 👉 Finger-Nibbling Exercise 3

Let's see what exponential growth looks like.

1. Change the range so the plot goes all the way to **256 bits**.
2. Run the cell and see what changes.
3. Now remove `plt.yscale("log")` and run it again.

**Think about it:** Why is the logarithmic scale so helpful for viewing this data?

There’s nothing to submit. Experiment with the plot and see what happens.

```python
import matplotlib.pyplot as plt

bits = range(8, 129, 8)
guesses_per_second = 1_000_000_000_000

years = [
    (2 ** b) / guesses_per_second / (60 * 60 * 24 * 365)
    for b in bits
]

plt.plot(bits, years, marker="o")
plt.yscale("log")

plt.xlabel("Secret size (bits)")
plt.ylabel("Years to try all possibilities")
plt.title("How Fast Does the Search Space Grow?")
plt.grid(True)
plt.show()
```
